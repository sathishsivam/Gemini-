```python
from transformers import AutoTokenizer,AutoModelForSeq2SeqLM
```

```python
tokenzier=AutoTokenizer.from_pretrained("facebook/nllb-200-distilled-600M")
```

```python
model=AutoModelForSeq2SeqLM.from_pretrained("facebook/nllb-200-distilled-600M")
```

```python
text="Swamiye Saranam Ayyappa, You are the one and only God to save the dharma, I can feel you from my heart and doing the pooja at all my instance"
```

```python
inputs=tokenzier(text,return_tensors="pt")
```

```python
translated_tokens=model.generate(**inputs,forced_bos_token_id=tokenzier.convert_tokens_to_ids("tam_Taml"),max_length=100)
```

```python
tokenzier.batch_decode(translated_tokens, skip_special_tokens=True)[0]
```
